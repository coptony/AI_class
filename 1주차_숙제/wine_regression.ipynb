```python
from google.colab import drive
drive.mount('/content/drive')

import pandas as pd

from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import RandomForestRegressor
from sklearn.linear_model import LinearRegression
from sklearn.neighbors import KNeighborsRegressor
from sklearn.metrics import mean_squared_error


# 데이터 불러오기
df = pd.read_csv("/content/drive/MyDrive/wine.csv")

# 기존 Classification 레이블인 Wine 제거
df = df.drop("Wine", axis=1)

# Alcohol을 Regression 정답(y)으로 설정
X = df.drop("Alcohol", axis=1)
y = df["Alcohol"]

# 학습용 / 테스트용 데이터 분리
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2
)


# 1. Decision Tree Regression
model = DecisionTreeRegressor()

model.fit(X_train, y_train)

pred_dt = model.predict(X_test)

print("===== Decision Tree Regression =====")
print("평균제곱오차:", mean_squared_error(y_test, pred_dt))


# 2. Random Forest Regression
model = RandomForestRegressor()

model.fit(X_train, y_train)

pred_rf = model.predict(X_test)

print("===== Random Forest Regression =====")
print("평균제곱오차:", mean_squared_error(y_test, pred_rf))


# 3. Linear Regression
model = LinearRegression()

model.fit(X_train, y_train)

pred_lr = model.predict(X_test)

print("===== Linear Regression =====")
print("평균제곱오차:", mean_squared_error(y_test, pred_lr))


# 4. KNN Regression
model = KNeighborsRegressor()

model.fit(X_train, y_train)

pred_knn = model.predict(X_test)

print("===== KNN Regression =====")
print("평균제곱오차:", mean_squared_error(y_test, pred_knn))
```
